```python
%sql
USE CATALOG mvp;
USE SCHEMA silver;

CREATE OR REPLACE TABLE comex_importacao_limpa 
USING delta
AS 
SELECT 
    CAST(CO_ANO AS INT) AS co_ano,
    CAST(CO_MES AS INT) AS co_mes,
   
    LPAD(CAST(CO_NCM AS STRING), 8, '0') AS co_ncm,

    SUBSTR(LPAD(CAST(CO_NCM AS STRING), 8, '0'), 1, 2) AS setor_capitulo,
    CAST(CO_PAIS AS INT) AS co_pais,
    UPPER(TRIM(SG_UF_NCM)) AS sg_uf_ncm,
  
    CAST(KG_LIQUIDO AS DOUBLE) AS kg_liquido,
    CAST(VL_FOB AS DOUBLE) AS vl_fob
FROM mvp.bronze.comex_importacao_total
WHERE CO_ANO IS NOT NULL 
  AND CO_PAIS IS NOT NULL 
  AND VL_FOB > 0;
```

```python
%sql
USE CATALOG mvp;
USE SCHEMA silver;


SELECT COUNT(*) AS total_linhas_silver FROM comex_importacao_limpa;


SELECT co_ano, co_ncm, setor_capitulo, co_pais, vl_fob, kg_liquido 
FROM comex_importacao_limpa 
LIMIT 10;
```
