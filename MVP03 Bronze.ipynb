```python
%sql
USE CATALOG mvp;
USE SCHEMA bronze;

CREATE TABLE comex_importacao_total
USING delta
AS 
SELECT *
FROM read_files ('/Volumes/mvp/staging/mdic/*.csv',
    format => 'csv',
    header => 'true',
    delimiter => ';',
    inferSchema => 'true')
```
